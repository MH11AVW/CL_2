```python
import io
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
stop_words = set(stopwords.words('english'))
```

```python
import os

os.makedirs("./data", exist_ok=True)
sample_path = "./data/sample.txt"
if not os.path.exists(sample_path):
    with open(sample_path, "w", encoding="utf-8") as f:
        f.write("Integrating non-traditional assessment methods like project-based learning and competency-based evaluation could provide a comprehensive picture of student performance. In conclusion, student performance monitoring systems offer immense potential for enhancing educational outcomes by leveraging data, technology, and collaboration.")

file1 = open(sample_path, encoding="utf-8")

line = file1.read()

words = line.split()

print(f"Before Removing Stop Words : \n{words}\nLength : {len(words)}")

file1.close()
```

```python
output_path = "./data/output_filteredtext.txt"
if os.path.exists(output_path):
    os.remove(output_path)

for r in words:
    if r.lower() not in stop_words:
        with open(output_path, "a", encoding="utf-8") as appendFile:
            appendFile.write(" " + r)
```

```python
file2 = open("./data/output_filteredtext.txt")

line2 = file2.read()

words = line2.split()

print(f"After Removing Stop Words : \n{words}\nLength : {len(words)}")
```
